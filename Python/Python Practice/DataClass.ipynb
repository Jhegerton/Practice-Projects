```python
from dataclasses import dataclass
```

```python
@dataclass
class Car:
    make: str
    model: str
    trim: str
    year: int

    def __str__(self) -> str:
        return f"{self.year} {self.make} {self.model} {self.trim}"
```

```python
my_car: Car = Car("Honda", "Civic", "LX", 2014)
```
