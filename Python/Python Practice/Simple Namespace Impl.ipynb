```python
class Namespace:
    """
    Manually implemented namespace class, analogue to types.SimpleNamespace
    """

    def __init__(self, **kwargs):
        self.__dict__.update(kwargs)

    def __repr__(self):
        return f"Namespace({', '.join(f'{k}={v!r}' for k, v in self.__dict__.items())})"

    def __eq__(self, other):
        if not isinstance(other, Namespace):
            return False
        return self.__dict__ == other.__dict__

    def __ne__(self, other):
        return not self.__eq__(other)

    def __getitem__(self, key):
        return self._to_dict()[key]

    def keys(self):
        return self._to_dict().keys()

    def _to_dict(self) -> dict:
        def convert(value):
            if isinstance(value, Namespace):
                return {k: convert(v) for k, v in value.__dict__.items()}
            elif isinstance(value, list):
                return [convert(item) for item in value]
            elif isinstance(value, tuple):
                return tuple(convert(item) for item in value)
            else:
                return value

        return convert(self)

    @classmethod
    def from_data(cls, d) -> "Namespace":
        def convert(value):
            if isinstance(value, dict):
                return cls(**{k: convert(v) for k, v in value.items()})
            elif isinstance(value, list):
                return [convert(item) for item in value]
            elif isinstance(value, tuple):
                return tuple(convert(item) for item in value)
            else:
                return value

        return convert(d)

    def copy(self):
        return Namespace(**self.__dict__)
```

```python
alice_dict: dict = {
    "name": "Alice",
    "aliases": ("Ally", "Lice"),
    "age": 30,
    "address": {"street": "123 Main St", "city": "Anytown", "zip": "12345"},
    "children": [{"name": "Bob", "age": 5}, {"name": "Charlie", "age": 3}],
}
alice = Namespace.from_data(alice_dict)
print(alice.name)  # Output: Alice
print(alice.aliases[1])  # Output: Lice
print(alice.age)  # Output: 30
print(alice.address.street)  # Output: 123 Main St
print(alice.address.city)  # Output: Anytown
print(alice.address.zip)  # Output: 12345
print(alice.children[0].name)  # Output: Bob
print(alice.children[0].age)  # Output: 5
print(alice.children[1].name)  # Output: Charlie
print(alice.children[1].age)  # Output: 3
```

```python
repr(
    alice
)  # Output: Namespace(name='Alice', age=30, address=Namespace(street='123 Main St', city='Anytown', zip='12345'))
```

```python
alice_dict_converted_back = dict(alice)
print(alice_dict_converted_back == alice_dict)  # Output: True
```

```python
alice_copy = alice.copy()
print(alice_copy == alice)  # Output: True
print(alice_copy is alice)  # Output: False
```

```python
try:
    alice.aliases[1] = "Lissy"
except TypeError as e:
    print(f"Error: {e}")
# Error: 'Namespace' object does not support item assignment
```
