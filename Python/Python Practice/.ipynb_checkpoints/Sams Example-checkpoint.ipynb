```python
class Menu():

    def __init__(self, answer, limit=10):
        self._answer = answer.lower()
        self._limit = limit

    def __call__(self):
        i = 0
        check = False
        while(not check and (i < self._limit)):
            if self._answer != input("Enter your guess: ").lower():
                i+=1
                print("Try again!")
            else:
                check = True
        else:
            print("You did it!")
```

```python
menu = Menu("Sam", 10)
```

```python
menu()
```
