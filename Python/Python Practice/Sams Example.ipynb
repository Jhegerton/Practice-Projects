```python
class MenuClass:

    def __init__(self, answer, limit=10):
        self._answer = answer.lower()
        self._limit = limit

    def __call__(self):
        i = 0
        check = False
        while not check and (i < self._limit):
            if self._answer != (guess := input("Enter your guess: ").lower()):
                i += 1
                print(f"{guess} is incorrect. Try again!")
            else:
                check = True
        else:
            print("You did it!")
```

```python
def menu_function(answer, limit=10):
    i = 0
    check = False
    while not check and (i < limit):
        answer = answer.lower()
        guess = input("Enter your guess: ").lower()

        if answer != guess:
            i += 1
            print(f"{guess} is incorrect. Try again!")
        else:
            check = True
    else:
        print("You did it!")
```

```python
c_menu = MenuClass("Sam", 10)
```

```python
c_menu()
```

```python
menu_function("Sam", 10)
```
