```python
import math
from os import linesep


def add(x: int) -> callable:
    def inner(y: int) -> int:
        return x + y

    return inner


def power(exponent: int) -> callable:
    def inner(base: float) -> float:
        return base**exponent

    return inner


def root(degree: int) -> callable:
    if degree <= 0:
        raise ValueError("Degree must be a positive integer.")

    def inner(n: float) -> float:
        if n < 0:
            raise ValueError("Target number must be non-negative.")
        return n ** (1 / degree)

    return inner


def base(base: float | str = "e") -> callable:
    if base == "e":
        base = math.e

    def inner(exponent: float) -> float:
        return base**exponent

    return inner


def logarithm(base: int | str = "e") -> callable:
    if base != "e":
        if base <= 1:
            raise ValueError("Base must be greater than 1.")

    def inner(n: float) -> float:
        if n < 1:
            raise ValueError("Target number must be 1 or greater.")
        elif base == "e":
            return math.log(n)
        elif base == 10:
            return math.log10(n)
        elif base == 2:
            return math.log2(n)
        else:
            # Change of base formula: log_base(n) = log(n) / log(base)
            return math.log(n) / math.log(base)

    return inner


def join_strings(separator: str) -> callable:
    def inner(*args: str) -> str:
        return separator.join(args)

    return inner


def print_sep(seperator: str) -> callable:
    def inner(*args: str) -> None:
        print(*args, sep=seperator)

    return inner
```

```python
println = print_sep(linesep)
```

```python
sum1 = add(5)(10)
diff1 = add(11)(-5)
println(sum1, diff1)  # Output: 15 6
```

```python
add_ten = add(10)
add_hundred = add(100)
println(add_ten(20), add_hundred(20))  # Output: 30 120
```

```python
sqrt = root(2)
cbrt = root(3)
println(sqrt(16), cbrt(27))  # Output: 4.0 3.0
```

```python
square = power(2)
cube = power(3)
println(square(5), cube(3))  # Output: 25.0 27
```

```python
pow_2 = base(2)
pow_3 = base(3)
pow_e = base()
println(pow_2(8), pow_3(4), pow_e(1))  # Output: 256 64 2.718281828459045
```

```python
log10 = logarithm(10)
log2 = logarithm(2)
log3 = logarithm(3)
ln = logarithm()
println(log10(100), log2(2048), log3(27), ln(math.e))  # Output: 2.0 11.0 3.0 1
```

```python
join_space = join_strings(" ")
join_comma = join_strings(",")
join_semicolon = join_strings(";")

lst1 = ["Hello", "World"]
lst2 = ["apple", "banana", "cherry"]
lst3 = ["one", "two", "three"]

str1 = join_space(*lst1)
str2 = join_comma(*lst2)
str3 = join_semicolon(*lst3)

println(
    str1, str2, str3
)  # Output: "Hello World", "apple,banana,cherry", "one;two;three"
```
