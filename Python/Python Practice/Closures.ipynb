```python
import math
from os import linesep


def add(x: int) -> callable:
    def inner(y: int) -> int:
        return x + y

    return inner


def logarithm(base: int) -> callable:
    if base <= 1:
        raise ValueError("Base must be greater than 1.")

    def log(n: float) -> float:
        if n < 1:
            raise ValueError("Target number must be 1 or greater.")
        elif base == "e":
            return math.log(n)
        else:
            return math.log(n, base)

    return log


def join_strings(separator: str) -> callable:
    def joiner(*args: str) -> str:
        return separator.join(args)

    return joiner
```

```python
sum1 = add(5)(10)
diff1 = add(11)(-5)
print(sum1, diff1, sep=linesep)  # Output: 15 6
```

```python
add_ten = add(10)
add_hundred = add(100)
print(add_ten(20), add_hundred(20), sep=linesep)  # Output: 30 120
```

```python
log10 = logarithm(10)
log2 = logarithm(2)
print(log10(100), log2(2048), sep=linesep)  # Output: 2.0 11.0
```

```python
join_space = join_strings(" ")
join_comma = join_strings(",")
join_semicolon = join_strings(";")

str1 = join_space("Hello", "World")
str2 = join_comma("apple", "banana", "cherry")
str3 = join_semicolon("one", "two", "three")

print(
    str1, str2, str3, sep=linesep
)  # Output: "Hello World", "apple,banana,cherry", "one;two;three"
```
