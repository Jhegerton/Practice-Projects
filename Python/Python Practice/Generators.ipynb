```python
my_gen = (x + x for x in range(10))

print(my_gen)  # <generator object <genexpr> at 0x7f8c8c8c8c8>
for _ in range(5):
    print(next(my_gen))  # 0, 2, 4, 6, 8

print(list(my_gen))  # [10, 12, 14, 16, 18]
```

```python
def my_gen_func():
    for x in range(10):
        yield x + x

my_gen_func_instance = my_gen_func()
for _ in range(5):
    print(next(my_gen_func_instance))  # 0, 2, 4, 6, 8
print(list(my_gen_func_instance))  # [10, 12, 14, 16, 18]
```
