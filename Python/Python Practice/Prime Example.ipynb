```python
from math import sqrt, ceil
from functools import reduce
from operator import mul
from tabulate import tabulate
```

```python
def is_prime(number):
    min_val = 2
    max_val = ceil(sqrt(number)) + 1
    total_range = range(min_val, max_val)

    for num in total_range:
        if number % num == 0:
            return False

    return True
```

```python
def get_prime_factors_rec(number_to_factor):
    factors = []
    min_val = 2
    max_val = ceil(sqrt(number_to_factor)) + 1
    total_range = range(min_val, max_val)

    if not is_prime(number_to_factor):
        for num in total_range:
            if number_to_factor % num == 0:
                factors.append(num)
                factors.extend(get_prime_factors_rec(number_to_factor // num))
                return [factor for factor in factors if factor != 1]
    return [number_to_factor]


def get_prime_factors(number_to_factor):
    """Pythonic wrapper for get_prime_factors_rec."""

    def inner():
        return get_prime_factors_rec(number_to_factor)

    # return the inner function as closure
    return inner()
```

```python
fact1 = get_prime_factors(100)
fact2 = get_prime_factors(500)
fact3 = get_prime_factors(1234)
```

```python
factors = [fact1, fact2, fact3]
products = [reduce(mul, lst) for lst in factors]
```

```python
results = dict()
for lst, prod in zip(factors, products):
    results[prod] = tuple(lst)

headers = ["Value", "Prime Factors"]
print(tabulate(results.items(), headers=headers, tablefmt="rounded_outline"))
```
