```python
from types import SimpleNamespace


def dict_to_namespace(dct: dict) -> SimpleNamespace:
    """Convert a dictionary to a SimpleNamespace."""

    def convert(
        dct: dict,
    ) -> SimpleNamespace:
        """Recursively convert nested dictionaries to SimpleNamespace."""
        for key, value in dct.items():
            if isinstance(value, dict):
                dct[key] = convert(value)
        return SimpleNamespace(**dct)

    return convert(dct)
```

```python
alice_dict: dict = {
    "name": "Alice",
    "age": 30,
    "address": {"street": "123 Main St", "city": "Anytown", "zip": "12345"},
}
alice = dict_to_namespace(alice_dict)
print(alice.name)  # Output: Alice
print(alice.age)  # Output: 30
print(alice.address.street)  # Output: 123 Main St
print(alice.address.city)  # Output: Anytown
print(alice.address.zip)  # Output: 12345
```

```python
repr(
    alice
)  # Output: namespace(name='Alice', age=30, address=namespace(street='123 Main St', city='Anytown', zip='12345'))
```
