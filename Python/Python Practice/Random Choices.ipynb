```python
from random import choices
from tabulate import tabulate

menu: str = r"""
Let's dig for some treasure! What would you like to do?
1. Dig for treasure
2. Quit
"""

dig_info: str = r"""
These are the odds of finding treasure:

20% chance for quartz
30% chance for pyrite
40% chance for iron
7% chance for silver
2% chance for gold
1% chance for diamond
"""

dig_instructions: str = """
How many times would you like to dig for treasure?
"""


def build_results(treasures: list[str]) -> str:
    results: dict[str, int] = {}
    for treasure in treasures:
        if treasure in results:
            results[treasure] += 1
        else:
            results[treasure] = 1

    return tabulate(
        results.items(), headers=["Treasure", "Amount(g)"], tablefmt="rounded_outline"
    )


def dig() -> None:
    max_tries = 3
    max_digs = 1000
    tries = 0
    decision = input(menu)

    while tries <= max_tries:
        if decision == "1":
            print(dig_info)
            times = int(input(dig_instructions))
            if times < 1 or times > max_digs:
                print(f"Please enter a number between 1 and {max_digs}.")
                tries += 1
                continue
            print(f"Digging for treasure for {times} times...")
            treasures = choices(
                population=["Quartz", "Pyrite", "Iron", "Silver", "Gold", "Diamond"],
                weights=[20, 30, 40, 7, 2, 1],
                k=times,
            )
            print(
                "You finished digging!",
                "",
                "Here are your results:",
                build_results(treasures),
                sep="\n",
            )
        elif decision == "2":
            print("Thanks for playing! Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1 or 2.")
            tries += 1


def main():
    dig()


if __name__ == "__main__":
    main()
```
