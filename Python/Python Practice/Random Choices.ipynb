```python
from random import choices
from tabulate import tabulate

menu: str = r"""
Let's dig for some treasure! What would you like to do?
1. Dig for treasure
2. Quit
"""

dig_info: str = r"""
These are the odds of finding treasure:

20% chance for quartz
30% chance for pyrite
40% chance for iron
7%  chance for silver
2%  chance for gold
1%  chance for diamond
"""

dig_instructions: str = """
How many times would you like to dig for treasure?
"""


def build_results(treasures: list[str]) -> str:
    results: dict[str, int] = {}
    for treasure in treasures:
        if treasure in results:
            results[treasure] += 1
        else:
            results[treasure] = 1

    results = dict(sorted(results.items(), key=lambda x: x[1], reverse=True))

    return tabulate(
        results.items(), headers=["Treasure", "Amount(g)"], tablefmt="rounded_outline"
    )


def dig() -> None:
    max_tries = 3
    max_digs = 1000
    tries = 0

    while tries < max_tries:
        decision = input(menu)
        match decision:
            case "1":
                print(dig_info)
                correct_input = False
                times = 0
                while not correct_input and tries < max_tries:
                    times = int(input(dig_instructions))

                    if times not in range(1, max_digs + 1):
                        print(
                            f"Error: {times} is not a valid number of tries."
                            "\n"
                            f"Please enter a number between 1 and {max_digs}."
                        )
                        tries += 1
                        continue
                    else:
                        correct_input = True
                        break
                else:
                    print("Error: Too many invalid attempts. Exiting the game.")
                    return

                print(f"Digging for treasure for {times} times...", end="\n\n")

                treasures = choices(
                    population=(
                        "Quartz",
                        "Pyrite",
                        "Iron",
                        "Silver",
                        "Gold",
                        "Diamond",
                    ),
                    weights=[20, 30, 40, 7, 2, 1],
                    k=times,
                )

                print(
                    "You finished digging! \n"
                    "Here are your results: \n"
                    f"{build_results(treasures)}",
                    end="\n\n",
                )

            case "2":
                print("Thanks for digging! Goodbye!")
                break

            case _:
                print("Error: Invalid choice. Please enter 1 or 2.")
                tries += 1

    else:
        print("Error: Too many invalid attempts. Exiting the game.")


def main():
    dig()


if __name__ == "__main__":
    main()
```
