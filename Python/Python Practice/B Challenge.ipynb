```python
# Given an integer array nums, return true if any value appears at least twice in the array;
# otherwise return false.

# Examples
# nums = [1,2,3,1] -> true
# nums = [1,2,3,4] -> false
# nums = [1,1,1,3,3,4,3,2,4,2] -> true


def has_repeat(lst):  # actual submission
    st = set(lst)
    lst2 = list(st)
    if len(lst) == len(lst2):
        return False
    else:
        return True


def has_repeat2(lst):
    dic = dict()
    for i, item in enumerate(lst):
        if item in dic.values():
            return True
        dic[i] = item
    return False


def has_repeat3(lst):
    lst2 = list()
    for item in lst:
        if item in lst2:
            return True
        lst2.append(item)
    return False
```

```python
has_repeat([1, 2, 3, "k", 7, "k"])
```

```python
has_repeat2([1, 2, 3, 4, "l"])
```

```python
has_repeat3([1, 1, 1, 3, 3, 4, 3, 2, 4, 2])
```
